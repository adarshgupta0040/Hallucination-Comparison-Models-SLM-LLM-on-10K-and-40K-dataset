```python
import re

def evaluate_true_false(expected, prediction):

    expected = str(expected).strip().lower()
    prediction = str(prediction).strip().lower()

    if expected not in ["true", "false"]:
        return None

    # Look for an explicit answer
    prediction = re.sub(
        r"^(the answer is|answer:|final answer:)\s*",
        "",
        prediction
    )

    if prediction.startswith("true"):
        return expected == "true"

    if prediction.startswith("false"):
        return expected == "false"

    # If there is no explicit answer, leave for manual evaluation
    return None
```

```python
import re

def extract_numbers(text):

    text = str(text)

    numbers = re.findall(
        r"-?\d+(?:\.\d+)?",
        text
    )

    if len(numbers) == 0:
        return None

    return float(numbers[-1])


def evaluate_math(expected, prediction):

    expected_num = extract_numbers(expected)
    predicted_num = extract_numbers(prediction)

    if expected_num is None or predicted_num is None:
        return None

    return expected_num == predicted_num
```

```python
def generate_tf_answer(model, tokenizer, question):

    messages = [
        {
            "role": "system",
            "content": (
                "You are a strict True/False classification system. "
                "Determine whether the statement is factually true or false. "
                "Respond with exactly one word: either TRUE or FALSE."
            )
        },
        {
            "role": "user",
            "content": f"Statement: {question}\n\nAnswer:"
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        # Increased max_new_tokens to 15 to prevent truncation if the model adds padding
        outputs = model.generate(
            **inputs,
            max_new_tokens=15,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    answer = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return answer.strip()
```

```python
import torch
import pandas as pd
import gc
import os
import re

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig
)

SYSTEM_PROMPT = """
You are a precise True/False classification system.

Determine whether the given statement is factually correct.

Rules:
1. If the statement is factually correct, answer TRUE.
2. If the statement is factually incorrect, answer FALSE.
3. Do not explain your answer.
4. Do not provide additional text.
5. Respond with exactly one word: TRUE or FALSE.
"""

models = {
    # SLM models
    "Qwen_0.5B": "Qwen/Qwen2.5-0.5B-Instruct",
    "SmolLM2_1.7B": "HuggingFaceTB/SmolLM2-1.7B-Instruct",
    "TinyLlama_1.1B": "TinyLlama/TinyLlama-1.1B-Chat-v1.0",

    # LLM models
    "Qwen_7B": "Qwen/Qwen2.5-7B-Instruct",
    "Mistral_7B": "mistralai/Mistral-7B-Instruct-v0.3",
    "Zephyr_7B": "HuggingFaceH4/zephyr-7b-beta"
}

print("GPU available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
def run_model(model_name, model_path, test_df):

    print("\nLoading model:", model_name)

    tokenizer = AutoTokenizer.from_pretrained(model_path)

    # Use 4-bit quantization for 7B models
    if "7B" in model_name:
        quant_config = BitsAndBytesConfig(
            load_in_4bit=True,
            bnb_4bit_quant_type="nf4",
            bnb_4bit_compute_dtype=torch.float16
        )

        model = AutoModelForCausalLM.from_pretrained(
            model_path,
            quantization_config=quant_config,
            device_map="auto"
        )

    else:
        model = AutoModelForCausalLM.from_pretrained(
            model_path,
            torch_dtype=torch.float16,
            device_map="auto"
        )

    model.eval()

    predictions = []

    for i, row in test_df.iterrows():

        messages = [
            {"role": "system", "content": SYSTEM_PROMPT},
            {
                "role": "user",
                "content": "Statement: " + str(row["question"])
            }
        ]

        prompt = tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        inputs = tokenizer(
            prompt,
            return_tensors="pt"
        ).to(model.device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=10,
                do_sample=False,
                pad_token_id=tokenizer.eos_token_id
            )

        generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]

        answer = tokenizer.decode(
            generated_tokens,
            skip_special_tokens=True
        ).strip()

        predictions.append({
            "model": model_name,
            "question_id": row["question_id"],
            "unique_id": row["unique_id"],
            "language": row["language"],
            "category": row["category"],
            "question": row["question"],
            "expected": row["expected"],
            "prediction": answer
        })

        if (len(predictions) % 20) == 0:
            print(
                model_name,
                "- Completed:",
                len(predictions),
                "/",
                len(test_df)
            )

    result_df = pd.DataFrame(predictions)

    # Save individual model results
    os.makedirs("/content/tf_results", exist_ok=True)

    result_df.to_csv(
        f"/content/tf_results/{model_name}.csv",
        index=False
    )

    print("Saved:", model_name)

    del model
    del tokenizer
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result_df
```

```python
all_tf_results = []

for model_name, model_path in models.items():

    result_df = run_model(
        model_name,
        model_path,
        tf_sample
    )

    all_tf_results.append(result_df)

    print("Finished:", model_name)

# Combine all model predictions
tf_results_df = pd.concat(
    all_tf_results,
    ignore_index=True
)

# Save combined results
tf_results_df.to_csv(
    "/content/TF_All_Models_Results.csv",
    index=False
)

print("\nAll models completed!")
print("Total predictions:", len(tf_results_df))

display(tf_results_df.head(10))
```

```python
import re
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Normalize expected labels
tf_results_df["expected_label"] = (
    tf_results_df["expected"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Improved extraction that is highly robust to conversational wrapping
def extract_label(response):
    if pd.isna(response):
        return "INVALID"

    text = str(response).strip().upper()

    # First check exact output
    if text in ["TRUE", "FALSE"]:
        return text

    # Search for the last or most explicit occurrence of TRUE/FALSE in the text
    matches = re.findall(r"\b(TRUE|FALSE)\b", text)
    if matches:
        # Return the last mentioned decision (handles 'it is not TRUE, so FALSE' or conversational padding)
        return matches[-1]

    return "INVALID"


tf_results_df["predicted_label"] = (
    tf_results_df["prediction"].apply(extract_label)
)

# Correctness
tf_results_df["correct"] = (
    tf_results_df["expected_label"] ==
    tf_results_df["predicted_label"]
)

# Model-wise evaluation
results = []

for model in tf_results_df["model"].unique():

    subset = tf_results_df[
        tf_results_df["model"] == model
    ].copy()

    total = len(subset)

    invalid = (
        subset["predicted_label"] == "INVALID"
    ).sum()

    valid = subset[
        subset["predicted_label"].isin(["TRUE", "FALSE"])
    ]

    if len(valid) > 0:

        y_true = valid["expected_label"]
        y_pred = valid["predicted_label"]

        precision = precision_score(
            y_true, y_pred,
            pos_label="TRUE",
            zero_division=0
        )

        recall = recall_score(
            y_true, y_pred,
            pos_label="TRUE",
            zero_division=0
        )

        f1 = f1_score(
            y_true, y_pred,
            pos_label="TRUE",
            zero_division=0
        )

    else:
        precision = 0
        recall = 0
        f1 = 0

    results.append({

        "Model": model,
        "Total": total,
        "Correct": int(subset["correct"].sum()),
        "Incorrect": int(
            ((~subset["correct"]) &
             (subset["predicted_label"] != "INVALID")).sum()
        ),
        "Invalid": int(invalid),

        "Accuracy (%)": round(
            subset["correct"].mean() * 100, 2
        ),

        "Valid Response (%)": round(
            (total - invalid) / total * 100, 2
        ),

        "Precision (%)": round(precision * 100, 2),
        "Recall (%)": round(recall * 100, 2),
        "F1-score (%)": round(f1 * 100, 2)
    })


metrics_df = pd.DataFrame(results)

display(metrics_df)
```

```python

import torch
import pandas as pd
import gc
import os

from transformers import AutoTokenizer, AutoModelForCausalLM
from sklearn.metrics import confusion_matrix

# ============================================================
# TINYLLAMA - 40K DATASET DIRECT LABEL SCORING
# ============================================================

MODEL_PATH = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tiny_tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)

tiny_model = AutoModelForCausalLM.from_pretrained(
    MODEL_PATH,
    torch_dtype=torch.float16,
    device_map="auto"
)

tiny_model.eval()

# TinyLlama prompt
def tiny_build_prompt(question):

    messages = [
        {
            "role": "system",
            "content": (
                "Classify the factual correctness of the statement. "
                "Return TRUE if it is factually correct. "
                "Return FALSE if it is factually incorrect. "
                "The statement can be in any language. "
                "Your answer must be exactly TRUE or FALSE."
            )
        },
        {
            "role": "user",
            "content": "Statement: " + str(question)
        }
    ]

    return tiny_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )


def tiny_score(prompt, candidate):

    prompt_ids = tiny_tokenizer(
        prompt,
        add_special_tokens=False,
        return_tensors="pt"
    )["input_ids"]

    full_ids = tiny_tokenizer(
        prompt + candidate,
        add_special_tokens=False,
        return_tensors="pt"
    )["input_ids"]

    prompt_len = prompt_ids.shape[1]

    if not torch.equal(
        prompt_ids[0],
        full_ids[0, :prompt_len]
    ):
        raise ValueError("TinyLlama tokenization mismatch")

    full_ids = full_ids.to(tiny_model.device)

    with torch.inference_mode():

        logits = tiny_model(full_ids).logits

        log_probs = torch.log_softmax(
            logits[:, :-1, :].float(),
            dim=-1
        )

        target_ids = full_ids[:, 1:]

        token_log_probs = log_probs.gather(
            2,
            target_ids.unsqueeze(-1)
        ).squeeze(-1)

        candidate_scores = token_log_probs[
            0, prompt_len - 1:
        ]

    return candidate_scores.mean().item()


# Run on same 200 questions
tiny40_results = []

for i, row in tf_sample.iterrows():

    prompt = tiny_build_prompt(row["question"])

    true_score = tiny_score(prompt, " TRUE")
    false_score = tiny_score(prompt, " FALSE")

    prediction = (
        "TRUE" if true_score > false_score else "FALSE"
    )

    tiny40_results.append({

        "model": "TinyLlama_1.1B_DirectScoring",
        "question_id": row["question_id"],
        "unique_id": row["unique_id"],
        "language": row["language"],
        "category": row["category"],
        "question": row["question"],
        "expected": row["expected"],
        "prediction": prediction,
        "true_score": true_score,
        "false_score": false_score,
        "score_margin": true_score - false_score
    })

    if (len(tiny40_results) % 20) == 0:
        print("TinyLlama completed:",
              len(tiny40_results), "/", len(tf_sample))

tiny40_df = pd.DataFrame(tiny40_results)

tiny40_df["expected_label"] = (
    tiny40_df["expected"].astype(str).str.strip().str.upper()
)

tiny40_df["correct"] = (
    tiny40_df["expected_label"] == tiny40_df["prediction"]
)

tiny40_accuracy = tiny40_df["correct"].mean() * 100

print("\nTINYLLAMA DIRECT SCORING ACCURACY:",
      round(tiny40_accuracy, 2), "%")

print("\nPrediction distribution:")
print(tiny40_df["prediction"].value_counts())

print("\nConfusion matrix:")
print(confusion_matrix(
    tiny40_df["expected_label"],
    tiny40_df["prediction"],
    labels=["TRUE", "FALSE"]
))

tiny40_language = (
    tiny40_df.groupby("language")
    .agg(
        Total=("correct", "count"),
        Correct=("correct", "sum"),
        Accuracy=("correct", "mean")
    )
    .reset_index()
)

tiny40_language["Accuracy (%)"] = (
    tiny40_language["Accuracy"] * 100
).round(2)

display(tiny40_language)

os.makedirs("/content/tf_results", exist_ok=True)

tiny40_df.to_csv(
    "/content/tf_results/TinyLlama_40K_Direct_Scoring.csv",
    index=False
)

tiny40_language.to_csv(
    "/content/tf_results/TinyLlama_40K_Direct_Scoring_Language.csv",
    index=False
)

del tiny_model
del tiny_tokenizer
gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
```

```python

import torch
import pandas as pd
import gc
import os

from transformers import AutoTokenizer, AutoModelForCausalLM
from sklearn.metrics import confusion_matrix

# ============================================================
# SMOLLM2 1.7B - 40K DATASET CUSTOM PROMPT SCORING
# ============================================================

MODEL_PATH = "HuggingFaceTB/SmolLM2-1.7B-Instruct"

smol_tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)

smol_model = AutoModelForCausalLM.from_pretrained(
    MODEL_PATH,
    torch_dtype=torch.float16,
    device_map="auto"
)

smol_model.eval()


def smol_build_prompt(question):

    messages = [
        {
            "role": "system",
            "content": (
                "You perform binary fact checking. "
                "Read the entire statement carefully. "
                "Identify its main factual claim. "
                "If the claim describes facts correctly, choose TRUE. "
                "If the claim describes facts incorrectly, choose FALSE. "
                "The statement may be in English, Hindi, Gujarati, "
                "Marathi, or Odia. Understand its meaning before deciding. "
                "Do not guess. Do not explain. "
                "Choose exactly one label."
            )
        },
        {
            "role": "user",
            "content": (
                "Check this factual statement:\n"
                + str(question) +
                "\nLabel:"
            )
        }
    ]

    return smol_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )


def smol_score(prompt, candidate):

    prompt_ids = smol_tokenizer(
        prompt,
        add_special_tokens=False,
        return_tensors="pt"
    )["input_ids"]

    full_ids = smol_tokenizer(
        prompt + candidate,
        add_special_tokens=False,
        return_tensors="pt"
    )["input_ids"]

    prompt_len = prompt_ids.shape[1]

    if not torch.equal(
        prompt_ids[0],
        full_ids[0, :prompt_len]
    ):
        raise ValueError("SmolLM2 tokenization mismatch")

    full_ids = full_ids.to(smol_model.device)

    with torch.inference_mode():

        logits = smol_model(full_ids).logits

        log_probs = torch.log_softmax(
            logits[:, :-1, :].float(),
            dim=-1
        )

        target_ids = full_ids[:, 1:]

        token_log_probs = log_probs.gather(
            2,
            target_ids.unsqueeze(-1)
        ).squeeze(-1)

        candidate_scores = token_log_probs[
            0, prompt_len - 1:
        ]

    return candidate_scores.mean().item()


smol40_results = []

for i, row in tf_sample.iterrows():

    prompt = smol_build_prompt(row["question"])

    true_score = smol_score(prompt, " TRUE")
    false_score = smol_score(prompt, " FALSE")

    prediction = (
        "TRUE" if true_score > false_score else "FALSE"
    )

    smol40_results.append({

        "model": "SmolLM2_1.7B_CustomPrompt",
        "question_id": row["question_id"],
        "unique_id": row["unique_id"],
        "language": row["language"],
        "category": row["category"],
        "question": row["question"],
        "expected": row["expected"],
        "prediction": prediction,
        "true_score": true_score,
        "false_score": false_score,
        "score_margin": true_score - false_score
    })

    if (len(smol40_results) % 20) == 0:
        print("SmolLM2 completed:",
              len(smol40_results), "/", len(tf_sample))

smol40_df = pd.DataFrame(smol40_results)

smol40_df["expected_label"] = (
    smol40_df["expected"].astype(str).str.strip().str.upper()
)

smol40_df["correct"] = (
    smol40_df["expected_label"] == smol40_df["prediction"]
)

smol40_accuracy = smol40_df["correct"].mean() * 100

print("\nSMOLLM2 CUSTOM PROMPT ACCURACY:",
      round(smol40_accuracy, 2), "%")

print("\nPrediction distribution:")
print(smol40_df["prediction"].value_counts())

print("\nConfusion matrix:")
print(confusion_matrix(
    smol40_df["expected_label"],
    smol40_df["prediction"],
    labels=["TRUE", "FALSE"]
))

smol40_language = (
    smol40_df.groupby("language")
    .agg(
        Total=("correct", "count"),
        Correct=("correct", "sum"),
        Accuracy=("correct", "mean")
    )
    .reset_index()
)

smol40_language["Accuracy (%)"] = (
    smol40_language["Accuracy"] * 100
).round(2)

display(smol40_language)

os.makedirs("/content/tf_results", exist_ok=True)

smol40_df.to_csv(
    "/content/tf_results/SmolLM2_1.7B_40K_CustomPrompt.csv",
    index=False
)

smol40_language.to_csv(
    "/content/tf_results/SmolLM2_1.7B_40K_CustomPrompt_Language.csv",
    index=False
)

del smol_model
del smol_tokenizer
gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ============================================================
# 1. SELECT FINAL SIX MODELS
# ============================================================

# Modified SLM results
tiny_modified = tiny40_df.copy()
qwen_modified = qwen40_df.copy()
smol_modified = smol40_df.copy()

# Original LLM results
llm_models = [
    "Qwen_7B",
    "Mistral_7B",
    "Zephyr_7B"
]

llm_results = tf_results_df[
    tf_results_df["model"].isin(llm_models)
].copy()

# Standardize model names and types
tiny_modified["model"] = "TinyLlama_1.1B"
qwen_modified["model"] = "Qwen_0.5B"
smol_modified["model"] = "SmolLM2_1.7B"

tiny_modified["model_type"] = "SLM"
qwen_modified["model_type"] = "SLM"
smol_modified["model_type"] = "SLM"

llm_results["model_type"] = "LLM"

# Merge all six models
final_modified_results = pd.concat(
    [
        tiny_modified,
        qwen_modified,
        smol_modified,
        llm_results
    ],
    ignore_index=True
)

# ============================================================
# 2. NORMALIZE LABELS
# ============================================================

def normalize_prediction(value):

    if pd.isna(value):
        return "INVALID"

    value = str(value).strip().upper()

    if value in ["TRUE", "FALSE"]:
        return value

    matches = re.findall(r"\b(TRUE|FALSE)\b", value)

    if matches:
        return matches[-1]

    return "INVALID"


final_modified_results["expected_label"] = (
    final_modified_results["expected"]
    .astype(str)
    .str.strip()
    .str.upper()
)

final_modified_results["predicted_label"] = (
    final_modified_results["prediction"]
    .apply(normalize_prediction)
)

# Invalid predictions count as incorrect
final_modified_results["correct"] = (
    (final_modified_results["expected_label"] ==
     final_modified_results["predicted_label"]) &
    (final_modified_results["predicted_label"].isin(
        ["TRUE", "FALSE"]
    ))
)

# ============================================================
# 3. MODEL NAME MAPPING
# ============================================================

model_names = {
    "TinyLlama_1.1B": "TinyLlama 1.1B",
    "Qwen_0.5B": "Qwen 2.5 0.5B",
    "SmolLM2_1.7B": "SmolLM2 1.7B",
    "Qwen_7B": "Qwen 2.5 7B",
    "Mistral_7B": "Mistral 7B",
    "Zephyr_7B": "Zephyr 7B"
}

final_modified_results["Model"] = (
    final_modified_results["model"].map(model_names)
)

selected_models = list(model_names.keys())

# ============================================================
# 4. VALIDATE ALL SIX MODELS
# ============================================================

print("MODEL-WISE PREDICTION COUNTS")

display(
    final_modified_results.groupby(
        ["model_type", "Model"]
    ).agg(
        Predictions=("prediction", "count"),
        Unique_Questions=("unique_id", "nunique")
    ).reset_index()
)

for model in selected_models:

    subset = final_modified_results[
        final_modified_results["model"] == model
    ]

    if len(subset) != 200:
        raise ValueError(
            model + " does not contain exactly 200 predictions."
        )

    reference_ids = set(tf_sample["unique_id"])
    model_ids = set(subset["unique_id"])

    if reference_ids != model_ids:
        raise ValueError(
            model + " does not use the same test questions."
        )

print("\nAll six models validated successfully!")

# ============================================================
# 5. INDIVIDUAL MODEL PERFORMANCE
# ============================================================

model_metrics = []

for model in selected_models:

    subset = final_modified_results[
        final_modified_results["model"] == model
    ]

    valid = subset[
        subset["predicted_label"].isin(["TRUE", "FALSE"])
    ]

    y_true = valid["expected_label"]
    y_pred = valid["predicted_label"]

    total = len(subset)
    correct = int(subset["correct"].sum())
    invalid = total - len(valid)

    model_metrics.append({

        "Model": model_names[model],
        "Type": subset["model_type"].iloc[0],

        "Total": total,
        "Correct": correct,
        "Incorrect": total - correct,
        "Invalid": invalid,

        "Accuracy (%)": round(
            correct / total * 100, 2
        ),

        "Balanced Accuracy (%)": round(
            balanced_accuracy_score(y_true, y_pred) * 100, 2
        ) if len(valid) else 0,

        "Precision (%)": round(
            precision_score(
                y_true, y_pred,
                pos_label="TRUE",
                zero_division=0
            ) * 100, 2
        ) if len(valid) else 0,

        "Recall (%)": round(
            recall_score(
                y_true, y_pred,
                pos_label="TRUE",
                zero_division=0
            ) * 100, 2
        ) if len(valid) else 0,

        "F1-score (%)": round(
            f1_score(
                y_true, y_pred,
                pos_label="TRUE",
                zero_division=0
            ) * 100, 2
        ) if len(valid) else 0
    })

model_metrics_df = pd.DataFrame(model_metrics)

print("\nINDIVIDUAL PERFORMANCE - SIX MODELS")

display(model_metrics_df)

# ============================================================
# 6. LANGUAGE-WISE ACCURACY - ALL SIX MODELS
# ============================================================

language_results = []

for model in selected_models:

    for language in sorted(
        final_modified_results["language"].unique()
    ):

        subset = final_modified_results[
            (final_modified_results["model"] == model) &
            (final_modified_results["language"] == language)
        ]

        if len(subset) == 0:
            continue

        language_results.append({

            "Model": model_names[model],
            "Type": subset["model_type"].iloc[0],
            "Language": language,
            "Total": len(subset),
            "Correct": int(subset["correct"].sum()),

            "Accuracy (%)": round(
                subset["correct"].mean() * 100, 2
            )
        })

language_accuracy_df = pd.DataFrame(language_results)

print("\nLANGUAGE-WISE ACCURACY - SIX MODELS")

display(
    language_accuracy_df.pivot(
        index=["Type", "Model"],
        columns="Language",
        values="Accuracy (%)"
    )
)

# ============================================================
# 7. LANGUAGE-WISE SLM VS LLM AVERAGE
# ============================================================

language_group_avg = (
    language_accuracy_df
    .groupby(["Type", "Language"])["Accuracy (%)"]
    .mean()
    .unstack()
    .round(2)
)

print("\nLANGUAGE-WISE SLM VS LLM AVERAGE")

display(language_group_avg)

# ============================================================
# 8. CONFUSION MATRICES - ALL SIX MODELS
# ============================================================

fig, axes = plt.subplots(2, 3, figsize=(15, 9))

axes = axes.flatten()

for i, model in enumerate(selected_models):

    subset = final_modified_results[
        final_modified_results["model"] == model
    ]

    valid = subset[
        subset["predicted_label"].isin(["TRUE", "FALSE"])
    ]

    cm = confusion_matrix(
        valid["expected_label"],
        valid["predicted_label"],
        labels=["TRUE", "FALSE"]
    )

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["TRUE", "FALSE"]
    )

    disp.plot(
        ax=axes[i],
        cmap="Blues",
        colorbar=False,
        values_format="d"
    )

    invalid_count = (
        subset["predicted_label"] == "INVALID"
    ).sum()

    axes[i].set_title(
        model_names[model] +
        "\nInvalid: " + str(invalid_count)
    )

plt.suptitle(
    "Final Six Model Confusion Matrices - 40K Dataset",
    fontsize=15
)

plt.tight_layout()
plt.show()

# ============================================================
# 9. FINAL SLM VS LLM COMPARISON
# AVERAGE OF THREE MODELS
# ============================================================

comparison_rows = []

for model_type in ["LLM", "SLM"]:

    group = model_metrics_df[
        model_metrics_df["Type"] == model_type
    ]

    comparison_rows.append({

        "Model Type": model_type,

        "Number of Models": len(group),

        "Total Predictions": int(group["Total"].sum()),

        "Correct Predictions": int(group["Correct"].sum()),

        # Arithmetic average of the three model accuracies
        "Average Accuracy (%)": round(
            group["Accuracy (%)"].mean(), 2
        ),

        "Average Balanced Accuracy (%)": round(
            group["Balanced Accuracy (%)"].mean(), 2
        ),

        "Average Precision (%)": round(
            group["Precision (%)"].mean(), 2
        ),

        "Average Recall (%)": round(
            group["Recall (%)"].mean(), 2
        ),

        "Average F1-score (%)": round(
            group["F1-score (%)"].mean(), 2
        ),

        "Total Invalid Predictions": int(
            group["Invalid"].sum()
        )
    })

final_comparison_df = pd.DataFrame(comparison_rows)

print("\n" + "=" * 70)
print("FINAL MODIFIED SLM VS LLM COMPARISON - 40K DATASET")
print("=" * 70)

display(final_comparison_df)

# ============================================================
# 10. SAVE FINAL RESULTS
# ============================================================

final_modified_results.to_csv(
    "/content/40K_Final_Modified_Six_Model_Results.csv",
    index=False
)

model_metrics_df.to_csv(
    "/content/40K_Final_Modified_Individual_Model_Metrics.csv",
    index=False
)

language_accuracy_df.to_csv(
    "/content/40K_Final_Modified_Language_Accuracy.csv",
    index=False
)

language_group_avg.to_csv(
    "/content/40K_Final_Modified_SLM_LLM_Language_Average.csv"
)

final_comparison_df.to_csv(
    "/content/40K_Final_Modified_SLM_vs_LLM.csv",
    index=False
)

print("\nAll final comparison files saved successfully!")
```
